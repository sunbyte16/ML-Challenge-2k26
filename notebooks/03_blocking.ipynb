# 03. Multi-Pass Blocking and Candidate Generation
Evaluates candidate recall and reduction ratio across multiple blocking keys.

```python
import sys
sys.path.append('../src')
import pandas as pd
from business_entity_resolution.config import config
from business_entity_resolution.data_loader import load_train_data, parse_ground_truth_mapping
from business_entity_resolution.preprocessing import preprocess_dataframe
from business_entity_resolution.blocking import MultiPassBlocker

train_data = load_train_data(config)
df_s1 = preprocess_dataframe(train_data['source1'], is_source1=True)
df_s2 = preprocess_dataframe(train_data['source2'], is_source1=False)
df_s3 = preprocess_dataframe(train_data['source3'], is_source1=False)
df_targets = pd.concat([df_s2, df_s3], ignore_index=True)
gt_map = parse_ground_truth_mapping(train_data['ground_truth'])

blocker = MultiPassBlocker()
cands = blocker.generate_candidates(df_s1, df_s2, df_s3)
metrics = blocker.evaluate_blocking(cands, gt_map, len(df_targets))
```

### Blocking Takeaways
- Multi-pass blocking unions exact names, 1-token prefixes, 2-token prefixes, composite keys, and TF-IDF fallback.
- Generates candidate pairs reducing Cartesian search space by >99% while preserving true matches.
